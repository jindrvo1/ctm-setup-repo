```python
import os
from typing import NamedTuple

import pandas as pd
import pyperclip

pd.set_option("display.max_columns", 500)
```

```python
class DataFrameFilter(NamedTuple):
    column: str
    value: str


def load_data(file_path: str, filter: DataFrameFilter | None = None) -> pd.DataFrame:
    df = pd.read_excel(file_path)

    if filter is None:
        return df

    assert filter.column in df.columns, (
        f"Column {filter.column} does not exist in the DataFrame"
    )
    df = df[df[filter.column] == filter.value]

    return df
```

```python
def print_grouped_df(df_grouped: pd.DataFrame | pd.Series) -> None:
    for course_name, df in df_grouped.items():
        print(f"{course_name}:")
        print(df)
        print()


def print_students(
    df: pd.DataFrame,
    cols: list[str] | None = None,
) -> None:
    if cols is None:
        cols = ["Student First", "Student Last", "Student Email", "Parent Email"]

    grouped = df.groupby("Course Name")[cols].apply(
        lambda row: "\t" + "\n\t".join(["; ".join(student) for student in row.values])
    )
    print_grouped_df(grouped)


def print_emails(df: pd.DataFrame, col: str) -> None:
    grouped = df.groupby("Course Name")[col].apply(lambda row: "\t" + "; ".join(row))
    print_grouped_df(grouped)


def print_student_emails(df: pd.DataFrame) -> None:
    print_emails(df, "Student Email")


def print_parent_emails(df: pd.DataFrame) -> None:
    print_emails(df, "Parent Email")
```

```python
data_folder = "data"
data_file = "1168-all-students-fall-2026.xlsx"
data_path = os.path.join(data_folder, data_file)


df = load_data(data_path, DataFrameFilter(column="CTM Instructor Last", value="Jindra"))
```

```python
print_students(df)
```

```python
print_student_emails(df)
```

```python
print_parent_emails(df)
```

```python
def handle_input(prompt: str) -> bool:
    return input(f"{prompt}\nWrite 'q' to quit.").strip().lower() != "q"


def generate_welcome_emails(df: pd.DataFrame) -> None:
    with open("welcome_mail_template.txt", "r") as f:
        mail_template_orig = f.read()

    firstname = "$firstname"
    course_name = "$course_name"
    username = "$username"
    password = "$password"

    for row_id, row in df.iterrows():
        student_id = row_id

        mail_template = mail_template_orig.replace(firstname, row["Student First"])
        mail_template = mail_template.replace(course_name, row["Course Name"])
        mail_template = mail_template.replace(username, row["Enrollment Username"])
        mail_template = mail_template.replace(password, row["Enrollment Password"])

        print(f"Student ID = {student_id}:")

        email_parts = [
            ("Subject copied.", f"Welcome to {row['Course Name']}"),
            ("Recipients copied.", f"{row['Student Email']};{row['Parent Email']}"),
            ("Message copied.", mail_template),
        ]
        for label, value in email_parts:
            pyperclip.copy(value)
            print(f"\t{label}")

            if not handle_input(label):
                return
```

```python
generate_welcome_emails(df)
```
